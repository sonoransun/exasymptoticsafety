# 03 — Gravity-matter NGFPs vs. number of scalars

Promote the scalar quartic `λ_φ` to a dynamical coupling and track how
the shifted NGFP evolves as we add scalars.

References: Buccio, Percacci et al. (2025); Eichhorn–Schiffer (2022).


```python
import numpy as np
from asymsafety.beta.matter import scan_gravity_matter_fps
from asymsafety.visualization.style import (
    apply_style, coupling_label,
    COLOR_RELEVANT, COLOR_IRRELEVANT, COLOR_NGFP,
)
import matplotlib.pyplot as plt
apply_style()
```

## 1. Run the scan


```python
results = scan_gravity_matter_fps(
    n_scalars_range=range(0, 9),
    scalar_quartic=True,
)
for r in results:
    if r['fp_exists']:
        print(f"N_s={r['n_scalars']}: g={r['g']:.4f}, λ={r['lambda']:.4f}, λ_φ={r['lambda_phi']:.4f}, n_relevant={r['n_relevant']}")
    else:
        print(f"N_s={r['n_scalars']}: NGFP not found from this guess")
```

## 2. Plot the shift


```python
found = [r for r in results if r['fp_exists']]
ns      = [r['n_scalars']  for r in found]
g_star  = [r['g']          for r in found]
l_star  = [r['lambda']     for r in found]
lp_star = [r['lambda_phi'] for r in found]

fig, axes = plt.subplots(1, 3, figsize=(13, 4))
for ax, ys, name, color in zip(
    axes,
    [g_star, l_star, lp_star],
    ['g', 'lambda', 'lambda_phi'],
    [COLOR_RELEVANT, COLOR_IRRELEVANT, COLOR_NGFP],
):
    ax.plot(ns, ys, 'o-', color=color, lw=2.0)
    ax.set_xlabel(coupling_label('n_s'))
    ax.set_ylabel(coupling_label(name) + r'$^{*}$')
    ax.grid(True, alpha=0.3)
fig.suptitle('Gravity-matter NGFP shift with $N_s$', fontsize=13)
fig.tight_layout()
```

Adding scalars typically pushes `g*` toward smaller values and `λ*`
toward the graviton pole at 1/2 — at some N_s the NGFP merges with
another and disappears, giving the scalar matter bound (Korver–
Saueressig–Wang 2024).
