# 02 — Phase portrait of the EH flow

Visualize the RG flow in the (g, λ) plane: streamlines, fixed points,
and a few representative trajectories integrated with
`scipy.integrate.solve_ivp`.


```python
import matplotlib.pyplot as plt
from asymsafety.beta.einstein_hilbert import build_eh_beta_system
from asymsafety.analysis.fixed_points import FixedPointFinder
from asymsafety.analysis.flow import FlowIntegrator
from asymsafety.visualization.phase_portrait import phase_portrait_2d
```

## 1. Build the system and find the NGFP


```python
system = build_eh_beta_system(d=4)
finder = FixedPointFinder(system)
ngfp = finder.find_fixed_point({'g': 0.7, 'lambda': 0.14})
gfp = finder.gaussian_fixed_point()
fps = [gfp, ngfp]
```

## 2. Integrate three trajectories

Each trajectory is integrated from a slightly displaced initial
condition near the NGFP toward the IR (decreasing t).


```python
flow = FlowIntegrator(system)
trajectories = []
for delta_g in (-0.02, 0.0, 0.02):
    ic = {'g': ngfp.location['g'] + delta_g, 'lambda': ngfp.location['lambda'] + 0.01}
    traj = flow.integrate(ic, t_span=(10, -10), max_step=0.05)
    trajectories.append(traj)
print(f'Built {len(trajectories)} trajectories.')
```

## 3. Plot the phase portrait


```python
fig = phase_portrait_2d(
    system,
    x_coupling='g',
    y_coupling='lambda',
    x_range=(-0.1, 1.5),
    y_range=(-0.1, 0.45),
    fixed_points=fps,
    trajectories=trajectories,
    title='EH flow in the (g, λ) plane',
)
plt.show()
```

The streamlines show the RG flow direction. Trajectories starting near
the NGFP funnel into either the GFP (separatrix) or the asymmetric IR
regions, depending on initial conditions.
